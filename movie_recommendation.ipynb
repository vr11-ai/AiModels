```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from google.colab import drive
drive.mount('/content/drive')

# 1. Load the datasets
movies = pd.read_csv('/content/drive/MyDrive/datasets/movies/movies.csv')
ratings = pd.read_csv('/content/drive/MyDrive/datasets/movies/ratings.csv')

# 2. Merge data to combine movie titles with their ratings
df = pd.merge(ratings, movies, on='movieId')

# 3. Create a User-Item Matrix
# This table shows every user's rating for every movie.
user_movie_matrix = df.pivot_table(index='userId', columns='title', values='rating').fillna(0)

# 4. Calculate Item-Item Similarity
item_similarity = cosine_similarity(user_movie_matrix.T)

# Convert the result into a DataFrame for easy lookup
item_similarity_df = pd.DataFrame(
    item_similarity,
    index=user_movie_matrix.columns,
    columns=user_movie_matrix.columns
)

# 5. Recommendation Function
def get_recommendations(movie_title, num_recommendations=10):
    """
    Takes a movie title and returns the top 'n' most similar movies.
    """
    if movie_title not in item_similarity_df.columns:
        return f"Error: '{movie_title}' not found in the dataset."

    # Get similarity scores for this movie against all others
    scores = item_similarity_df[movie_title].sort_values(ascending=False)

    # Return top results (excluding the first one, which is the movie itself)
    return scores.iloc[1:num_recommendations+1]

# 6. Usage Examples
print("--- Recommendation System Ready ---")

# Example: Recommendations for 'Toy Story (1995)'
target_movie = "Toy Story (1995)"
print(f"\nTop 10 Recommendations for users who liked '{target_movie}':")
print(get_recommendations(target_movie))

# Example: Recommendations for 'Fight Club (1999)'
target_movie2 = "Fight Club (1999)"
print(f"\nTop 10 Recommendations for users who liked '{target_movie2}':")
print(get_recommendations(target_movie2))
```
